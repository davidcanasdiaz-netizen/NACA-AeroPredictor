```python
import joblib
import numpy as np

def predecir_manual_capa_limite(delta, U_e, u_tau, Pi_w, delta_star, theta_star, Tu, U):
    # Cargar los archivos
    scaler = joblib.load('scaler_BL.pkl')
    ensamble = joblib.load('modelo_BL.pkl')

    modelo_delta = ensamble['modelo_delta']
    modelo_delta_star = ensamble['modelo_delta_star']
    modelo_theta_star = ensamble['modelo_theta_star']

    # Variables de entrada
    X_manual = np.array([[delta, U_e, u_tau, Pi_w, delta_star, theta_star, Tu, U]])

    X_scaled = scaler.transform(X_manual)

    # Prediccion
    pred_delta = modelo_delta.predict(X_manual)[0]
    pred_delta_star = modelo_delta_star.predict(X_scaled)[0]
    pred_theta_star = modelo_theta_star.predict(X_manual)[0]

    return pred_delta, pred_delta_star, pred_theta_star

# Ingreso de las variables

delta      = 0.0081     # Espesor de la capa límite
U_e        = 26.7947    # Velocidad en el borde de la capa límite
u_tau      = 0.9851     # Velocidad de cizalladura
Pi_w       = 1.0991     # Factor de estela
delta_star = 0.000885   # Espesor de desplazamiento
theta_star = 0.000694   # Espesor de cantidad de movimiento
Tu         = 13      # Intensidad de turbulencia de entrada
U          = 20.0       # Velocidad del flujo externo

# Ejecutar predicción
resultado_delta, resultado_delta_star, resultado_theta_star = predecir_manual_capa_limite(
    delta, U_e, u_tau, Pi_w, delta_star, theta_star, Tu, U
)

print("Predicción")
print(f"Ratio de Espesor (Rδ)         : {resultado_delta:.4f}")
print(f"Ratio de Desplazamiento (Rδ*) : {resultado_delta_star:.4f}")
print(f"Ratio de Momentum (Rθ*)       : {resultado_theta_star:.4f}")
```
